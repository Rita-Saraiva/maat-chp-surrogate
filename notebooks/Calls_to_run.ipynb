### Small Calls

```python
objectives = ["Econ", "QoL"]


# datasets per objective
datasets = {
    "Econ": ["econ_combined"] ,
    "QoL":  ["qol_combined"],
}

# single-head components (== old "targets") per objective
components = {
    "Econ": ["delay", "cancel" , "infra"],
    "QoL":  ["qol", "action", "maintenance"],
}

# profiles (== PARAMETERS) per model family
profiles = {
    "MLP":  ["simple_mlp"],
    "Auto": ["small_bottleneck", "medium_bottleneck"],
    "VAE":  ["small_vae", "medium_vae"],
    "GP":   ["tiny_gp"], 
    "XGB":  ["small_xgb"],
    "GNN": ["small_gnn"]
}

```

## Single

```python
for objective in objectives:
    for dataset in datasets[objective]:
        for model, model_profiles in profiles.items():
            for parameter in model_profiles:
                for component in components[objective]:
                    # for frac in [0.1, 0.25, 0.5]:
                        train_fraction = 0.25
                        print(
                            # f'TRAIN_FRACTION={train_fraction} '
                            # f'STAGE_TO_TMP="1" '
                        f'OBJECTIVE="{objective}" DATASET="{dataset}" MODE="single" '
                        f'MODEL="{model}" PARAMETERS="{parameter}" COMPONENT="{component}" '
                        # f'qsub -v TRAIN_FRACTION,STAGE_TO_TMP,OBJECTIVE,DATASET,MODE,MODEL,PARAMETERS,COMPONENT '
                        # f'qsub -v STAGE_TO_TMP,OBJECTIVE,DATASET,MODE,MODEL,PARAMETERS,COMPONENT '
                        f'qsub -v OBJECTIVE,DATASET,MODE,MODEL,PARAMETERS,COMPONENT '
                        f'jobs/train_surrogate.pbs'
                    )

```
